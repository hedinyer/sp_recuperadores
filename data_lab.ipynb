```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', lambda x: f'{x:,.0f}')
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

print("✅ Librerías cargadas")
```

```python
# 📁 Ruta del archivo (ajusta si está en otra carpeta)
archivo = "reporte_clientes_2026-05-14.csv"

# 📖 Leer CSV con configuración para Excel en español
df = pd.read_csv(
    archivo, 
    sep=';',              # Delimitador que usamos
    encoding='utf-8-sig', # Para tildes y ñ
    decimal=',',          # Si los números usan coma decimal (opcional)
    thousands='.'         # Si los miles usan punto (opcional)
)

print(f"✅ Archivo cargado: {df.shape[0]:,} filas × {df.shape[1]} columnas")
```
